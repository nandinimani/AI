```python
import random
def get_conflicts(board):
    n = len(board)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == j - i:
                conflicts += 1
    return conflicts

def hill_climbing_n_queens(n=8):
    restart_count = 0
    while True:
        restart_count += 1
        print("\nStarting with a new random board")
        board = [random.randint(0, n - 1) for i in range(n)]
        current_conflicts = get_conflicts(board)
        print("Initial board conflicts:",current_conflicts)
        
        while True:
            best_neighbor = list(board)
            best_conflicts = current_conflicts
            for col in range(n):
                original_row = board[col]
                for row in range(n):
                    if row == original_row:
                        continue
                    board[col] = row
                    new_conflicts = get_conflicts(board)
                    if new_conflicts < best_conflicts:
                        best_conflicts = new_conflicts
                        best_neighbor = list(board)
                board[col] = original_row
            if best_conflicts >= current_conflicts:
                if current_conflicts > 0:
                    print("INVALID!",current_conflicts,"conflicts.")
                break
            board = best_neighbor
            current_conflicts = best_conflicts
            print("Move the queen. Conflicts:",current_conflicts)
        if current_conflicts == 0:
            print("Optimal solution found after",restart_count,"restart.")
            return board

def print_board(board):
    n = len(board)
    print("\nFinal",n,"Queens Board Layout")
    for row in range(n):
        row_string = ""
        for col in range(n):
            if board[col] == row:
                row_string += "Q "
            else:
                row_string += ". "
        print(row_string)
n_queens = 4
solution = hill_climbing_n_queens(n_queens)
print_board(solution)
```
