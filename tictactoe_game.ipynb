```python
board = []
taken = []

for i in range(1, 10):
    board.append(i)

for i in board:
    if i % 3 == 0:
        print(i, "\n")
    else:
        print(i, end=" ")

print("Human starts! Place X at any position from 1-9")

while True:
    X = int(input("\nEnter position to play X: "))

    if X < 1 or X > 9:
        print("Invalid position, RETRY!")
        continue

    if X in taken:
        print("Position taken, RETRY!")
        continue

    taken.append(X)
    board[X - 1] = "X"

    for i in range(9):
        print(board[i], end=" ")
        if (i + 1) % 3 == 0:
            print()
            
    winning_positions = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    won = False

    for a, b, c in winning_positions:
        if board[a] == "X" and board[b] == "X" and board[c] == "X":
            won = True
            break

    if won:
        print("Human wins!")
        break

    if len(taken) == 9:
        print("It's a draw!")
        break

    for position in range(1, 10):
        if position not in taken:
            computer = position
            break
    
    taken.append(computer)
    board[computer - 1] = "O"
    
    print("\nComputer played O at position", computer)
    
    for i in range(9):
        print(board[i], end=" ")
        if (i + 1) % 3 == 0:
            print()

    won = False

    for a, b, c in winning_positions:
        if board[a] == "O" and board[b] == "O" and board[c] == "O":
            won = True
            break

    if won:
        print("Computer wins!")
        break

    if len(taken) == 9:
        print("It's a draw!")
        break
```
