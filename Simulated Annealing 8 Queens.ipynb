```python

import random
import math

N = 8

def conflicts(board):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if (board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j)):
                count += 1
    return count

def display(board):
    for row in range(N):
        print(" ".join("Q" if board[col] == row else "." for col in range(N)))

def run_simulated_annealing():
    board = [random.randint(0, 7) for _ in range(N)]
    T = 100.0
    alpha = 0.95
    step = 1
    print("Initial Board:")
    display(board)
    
    while T > 0.01:
        current = conflicts(board)
        if current == 0:
            return board

        new_board = board.copy()
        col = random.randint(0, 7)
        new_row = random.randint(0, 7)

        while new_row == board[col]:
            new_row = random.randint(0, 7)

        new_board[col] = new_row
        new_conflicts = conflicts(new_board)
        delta = new_conflicts - current

        print("\n------------------------------")
        print("Iteration:", step)
        print("Temperature:", round(T, 2))
        print("Neighbour Board:")
        display(new_board)
        print("Old conflicts:", current)
        print("New conflicts:", new_conflicts)
        print("Delta E:", delta)

        if delta <= 0 or random.random() < math.exp(-delta / T):
            board = new_board
            print("Result: ACCEPTED")
        else:
            print("Result: REJECTED")

        T *= alpha
        step += 1

    return board if conflicts(board) == 0 else None


def simulated_annealing():
    attempt = 1

    while True:
        print("\n========== ATTEMPT", attempt, "==========")
        solution = run_simulated_annealing()
        if solution is not None:
            print("\n_______________________________")
            print("         FINAL ANSWER")
            print("Solution found!")
            print("Final conflicts:", conflicts(solution))
            display(solution)
            break
        else:
            print("\nSolution not found. Restarting...")
            attempt += 1


simulated_annealing()
```
