```python
room_a = input("Enter status of Room A (DIRTY/CLEAN): ")
room_b = input("Enter status of Room B (DIRTY/CLEAN): ")
position = input("Enter vacuum position (A/B): ")

print("\nInitial State:")
print(f"Room A: {room_a}, Room B: {room_b}, Position: {position}")
print()

if position == "A":
    if room_a == "DIRTY":
        print("Suck")
        room_a = "CLEAN"

    if room_b == "DIRTY":
        print("Move Right")
        position = "B"
        print("Suck")
        room_b = "CLEAN"

elif position == "B":
    if room_b == "DIRTY":
        print("Suck")
        room_b = "CLEAN"

    if room_a == "DIRTY":
        print("Move Left")
        position = "A"
        print("Suck")
        room_a = "CLEAN"

print("\nFinal State:")
print(f"Room A: {room_a}, Room B: {room_b}")
print(f"Vacuum Position: {position}")
```
