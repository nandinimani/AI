```python
import heapq

# Take input
initial = tuple(map(int, input("Enter initial state (0 for blank): ").split()))
goal = tuple(map(int, input("Enter goal state (0 for blank): ").split()))


def heuristic(state):
    h = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            h += 1

    return h


def print_state(state):
    for i in range(9):
        print(state[i], end=" ")

        if i % 3 == 2:
            print()


def get_neighbors(state):

    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up
    if row > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 3] = \
            new_state[blank - 3], new_state[blank]

        neighbors.append(tuple(new_state))

    # Down
    if row < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 3] = \
            new_state[blank + 3], new_state[blank]

        neighbors.append(tuple(new_state))

    # Left
    if col > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 1] = \
            new_state[blank - 1], new_state[blank]

        neighbors.append(tuple(new_state))

    # Right
    if col < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 1] = \
            new_state[blank + 1], new_state[blank]

        neighbors.append(tuple(new_state))

    return neighbors

def a_star(initial, goal):

    open_list = []

    g_initial = 0
    h_initial = heuristic(initial)
    f_initial = g_initial + h_initial

    heapq.heappush(open_list, (f_initial, g_initial, initial))

    parent = {}

    g_cost = {initial: 0}

    visited = set()

    steps = []

    while open_list:

        f, g, current = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)

        h = heuristic(current)

        steps.append((current, g, h, f))

        if current == goal:

            for step_no, (state, g, h, f) in enumerate(steps):

                print("\nStep", step_no)
                print_state(state)
                print()
                print("g(n) =", g)
                print("h(n) =", h)
                print("f(n) = g(n) + h(n) =", f)

            print("Total moves =", g)
           

            return
        for neighbor in get_neighbors(current):

            if neighbor in visited:
                continue

            new_g = g + 1
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                h_neighbor = heuristic(neighbor)
                f_neighbor = new_g + h_neighbor

                parent[neighbor] = current

                heapq.heappush(
                    open_list,
                    (f_neighbor, new_g, neighbor)
                )

    print("No solution found.")


# -----------------------------
# Run A*
# -----------------------------
a_star(initial, goal)
```
