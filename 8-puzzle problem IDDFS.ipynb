```python
def depth_limited_dfs(state, goal, depth, path, visited):
    if state == goal:
        return path + [state]
    if depth == 0:
        return None
    state_tuple = tuple(state)
    visited.add(state_tuple)
    zero = state.index(0)
    row = zero // 3
    col = zero % 3
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero = new_row * 3 + new_col
            new_state = state.copy()
            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]
            new_state_tuple = tuple(new_state)
            if new_state_tuple not in visited:
                result = depth_limited_dfs(
                    new_state,
                    goal,
                    depth - 1,
                    path + [state],
                    visited
                )
                if result:
                    return result
    return None
    
def iddfs(initial, goal):
    depth = 0
    while True:
        visited = set()

        solution = depth_limited_dfs(
            initial,
            goal,
            depth,
            [],
            visited
        )
        if solution:
            return solution
        depth += 1
initial_state = list(map(int, input("Enter initial state: ").split()))

goal_state = [1, 2, 3, 4, 5, 6, 7, 8, 0]


solution = iddfs(initial_state, goal_state)

if solution:
    print("\nSolution:")

    for state in solution:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("No solution found")
```
